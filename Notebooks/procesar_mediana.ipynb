```python
import os
import cv2
from google.colab import drive

# =====================================================
# CONECTAR GOOGLE DRIVE
# =====================================================
drive.mount('/content/drive')

# Carpeta original
RUTA_ORIGINAL = "/content/drive/MyDrive/Proyecto_CNN/flowers"

# Carpeta destino
RUTA_DESTINO = "/content/drive/MyDrive/Proyecto_CNN/flowers_mediana"

# Clases
clases = ['daisy', 'dandelion', 'rose', 'sunflower', 'tulip']

# Tamaño del filtro de mediana
KERNEL_MEDIANA = 5

# =====================================================
# PROCESAR IMÁGENES
# =====================================================

total = 0

for clase in clases:

    carpeta_entrada = os.path.join(RUTA_ORIGINAL, clase)
    carpeta_salida = os.path.join(RUTA_DESTINO, clase)

    os.makedirs(carpeta_salida, exist_ok=True)

    print(f"Procesando: {clase}")

    for archivo in os.listdir(carpeta_entrada):

        if archivo.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):

            ruta_entrada = os.path.join(carpeta_entrada, archivo)

            # Leer imagen a color
            imagen = cv2.imread(ruta_entrada)

            if imagen is None:
                print("Error:", archivo)
                continue

            # Filtro de mediana
            imagen_filtrada = cv2.medianBlur(
                imagen,
                KERNEL_MEDIANA
            )

            ruta_salida = os.path.join(
                carpeta_salida,
                archivo
            )

            cv2.imwrite(
                ruta_salida,
                imagen_filtrada
            )

            total += 1

print(f"\nTotal de imágenes procesadas: {total}")
print("Proceso terminado.")
```
